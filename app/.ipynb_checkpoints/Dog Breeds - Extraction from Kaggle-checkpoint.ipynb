```python
# Imports
import os
import requests
import pandas as pd
from dotenv import load_dotenv
import json
import re
```

```python
import kagglehub
from kagglehub import KaggleDatasetAdapter

# Download latest version
path = kagglehub.dataset_download("prajwaldongre/top-dog-breeds-around-the-world")

print("Path to dataset files:", path)
```

```python
# Load dataset
df = pd.read_csv(path + "\\Dog Breads Around The World.csv")
print(df.shape)
print(df.info())
```

```python
df.head()
```

```python
# Check for duplicates
print("Duplicated names:", df["Name"].duplicated().sum())

duplicates = df[df["Name"].duplicated(keep=False)]

print(duplicates["Name"])
```

```python
# Drop duplicates
df = df.drop_duplicates(subset="Name", keep="first")
```

```python
# Check for null values
missing = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_pct": df.isnull().mean() * 100})

missing = missing.sort_values("missing_pct", ascending=False)
print(missing)
```

```python
print(df[["Name", "Origin", "Type"]].nunique())
print(df.dtypes)
```

```python
df.shape
```

```python
# Inspecting temperament
temperaments = (df["Unique Feature"].dropna())
print("Unique temperament terms:", temperaments.nunique())
print(temperaments.value_counts())
```
